# Notebook 2 · Hệ thức truy hồi: lập và giải
Đối chiếu công thức tường minh với dãy tính từ hệ thức, và hệ thức lập được với đếm trực tiếp.

```python
import itertools
from fractions import Fraction as F

def day(c,a0,N):
    a=list(a0)
    for n in range(len(c),N+1): a.append(sum(ci*a[n-1-i] for i,ci in enumerate(c)))
    return a

# đề 2017–2020: a_n = -14 a_{n-1} - 49 a_{n-2}, a0=3, a1=35 => (3-8n)(-7)^n
a=day([-14,-49],[3,35],12)
assert all(a[n]==(3-8*n)*(-7)**n for n in range(13)); print(a[:6])

# bậc 3 (đề trắc nghiệm): a_{n+3}=14a_{n+2}-59a_{n+1}+70a_n
a=day([14,-59,70],[-7,-20,-100],12)
assert all(a[n]==-7*2**n+3*5**n-3*7**n for n in range(13)); print(a[:6])

# 2.1.16a: a_n = 2a_{n-1}+a_{n-2}-2a_{n-3}
a=day([2,1,-2],[3,6,0],12); print(a[:8])
```

## Lập truy hồi: xâu nhị phân chứa 3 bit 1 liên tiếp

```python
def dem(n): return sum(1 for s in itertools.product((0,1),repeat=n) if '111' in ''.join(map(str,s)))
b={n:dem(n) for n in range(1,13)}
assert b[1]==b[2]==0 and b[3]==1
assert all(b[n]==b[n-1]+b[n-2]+b[n-3]+2**(n-3) for n in range(4,13))
print([b[n] for n in range(1,9)])   # a5 = 8

# từ mã thập phân có số lẻ chữ số 6: a_n = 8 a_{n-1} + 10^(n-1)
c={n:sum(1 for x in range(10**n) if str(x).zfill(n).count('6')%2==1) for n in range(1,5)}
assert c[1]==1 and all(c[n]==8*c[n-1]+10**(n-1) for n in range(2,5)); print(c)
```
