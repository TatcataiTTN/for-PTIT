# Notebook 1 · Đếm: bù trừ, nghiệm nguyên có cận, Dirichlet
Mỗi công thức trong Module 4–6 được đối chiếu với **đếm trực tiếp** (vét cạn) để chứng tỏ đáp án không gõ tay.

```python
import math, itertools

def dem_chia_het(a,b,ds):
    return sum(1 for x in range(a,b+1) if any(x%d==0 for d in ds))

def bu_tru(a,b,ds):
    tot=0
    for m in range(1,1<<len(ds)):
        sel=[ds[i] for i in range(len(ds)) if m>>i&1]; l=1
        for d in sel: l=l*d//math.gcd(l,d)
        c=b//l-(a-1)//l
        tot+= c if len(sel)%2 else -c
    return tot

for a,b,ds in [(1,5000,[6,9,15]),(542,7761,[3,7,14]),(5000,9999,[8,12]),(1000,5000,[6,9])]:
    print(ds,(a,b),bu_tru(a,b,ds),dem_chia_het(a,b,ds))
    assert bu_tru(a,b,ds)==dem_chia_het(a,b,ds)
```

## Nghiệm nguyên có cận (đề 2019–2020, câu 3a)
x₁+…+x₆ = 30 với 8 ≥ x₂ ≥ 3 và 6 ≥ x₄ ≥ 2 (kết quả đáp án viết tay trên ảnh đề: 58500).

```python
def dp_nghiem(N,lo,hi):
    dp=[1]+[0]*N
    for l,h in zip(lo,hi):
        nd=[0]*(N+1)
        for t in range(N+1):
            if dp[t]:
                for v in range(l,(N-t if h is None else min(h,N-t))+1): nd[t+v]+=dp[t]
        dp=nd
    return dp[N]

def bu_tru_nghiem(N,lo,hi):
    k=len(lo); base=N-sum(lo); idx=[i for i in range(k) if hi[i] is not None]; tot=0
    for m in range(1<<len(idx)):
        sel=[idx[j] for j in range(len(idx)) if m>>j&1]
        rem=base-sum(hi[i]-lo[i]+1 for i in sel)
        if rem>=0: tot+=(-1)**len(sel)*math.comb(rem+k-1,k-1)
    return tot

lo=[0,3,0,2,0,0]; hi=[None,8,None,6,None,None]
print(bu_tru_nghiem(30,lo,hi),dp_nghiem(30,lo,hi))
assert bu_tru_nghiem(30,lo,hi)==dp_nghiem(30,lo,hi)==58500
```
