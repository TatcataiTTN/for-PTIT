```python
import itertools, math
from fractions import Fraction as F

def hoan_vi_ke_tiep(p):
    p=list(p); i=len(p)-2
    while i>=0 and p[i]>p[i+1]: i-=1
    if i<0: return None
    j=len(p)-1
    while p[j]<p[i]: j-=1
    p[i],p[j]=p[j],p[i]; p[i+1:]=reversed(p[i+1:]); return tuple(p)

p=(5,6,8,3,9,7,4,2,1); out=[]
for _ in range(4): p=hoan_vi_ke_tiep(p); out.append(p)
print(out)   # 4 hoán vị kế tiếp của 568397421 (đề 2023–2024)
all_p=list(itertools.permutations(range(1,10))); i=all_p.index((5,6,8,3,9,7,4,2,1))
assert out==all_p[i+1:i+5]
```

```python
def nhanh_can(v,w,b):
    n=len(v); ord_=sorted(range(n),key=lambda i:(-F(v[i],w[i]),i)); C=[v[i] for i in ord_]; A=[w[i] for i in ord_]
    best=[-1,None]; nodes=[]
    def rec(k,delta,rem,x):
        for j in range(min(1,rem//A[k]),-1,-1):
            d2=delta+C[k]*j; r2=rem-A[k]*j; xx=x+[j]
            g=d2+F(C[k+1]*r2,A[k+1]) if k<n-1 else d2
            if k==n-1:
                if d2>best[0]: best[0]=d2; best[1]=xx
            elif g>best[0]: rec(k+1,d2,r2,xx)
            else: nodes.append((xx,float(g)))   # bị cắt
    rec(0,0,b,[]); xo=[0]*n
    for pos,orig in enumerate(ord_): xo[orig]=best[1][pos]
    return best[0],xo,nodes

print(nhanh_can([3,5,7,4],[2,4,3,2],9))   # đề 2023–2024, đề 01
for v,w,b in [([3,5,7,4],[2,4,3,2],9),([6,5,3,1],[4,4,2,1],9),([5,5,3,8],[3,4,2,5],12),([5,2,7,1],[5,3,6,4],9)]:
    bf=max(sum(a*x for a,x in zip(v,xs)) for xs in itertools.product((0,1),repeat=4) if sum(a*x for a,x in zip(w,xs))<=b)
    assert nhanh_can(v,w,b)[0]==bf; print((v,w,b),bf)
```
